# Social Media EDA

This notebook explores engagement patterns in the cleaned social media dataset.


```python
import pandas as pd
from config import SAMPLE_DATA_FILE
from src.data_loader import load_dataset
from src.data_cleaning import clean_social_media_data
from src.feature_engineering import add_engagement_features

df = load_dataset(SAMPLE_DATA_FILE)
cleaned_df, report = clean_social_media_data(df, source_name='sample')
df = add_engagement_features(cleaned_df)
df.head()
```

```python
df.describe(include='all')
```

```python
df.groupby('platform')['total_engagements'].mean().sort_values(ascending=False)
```

```python
df.groupby('content_type')['total_engagements'].mean().sort_values(ascending=False)
```

```python
df['post_date'] = pd.to_datetime(df['post_date'])
df.groupby(df['post_date'].dt.day_name())['total_engagements'].mean().reindex(['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'])
```
